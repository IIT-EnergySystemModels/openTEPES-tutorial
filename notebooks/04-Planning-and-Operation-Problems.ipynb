# Planning and Operation Problems

openTEPES is built from blocks, so the same model solves several different optimization problems. You choose which one by switching parameters on or off in the input CSV files.

The problems below range from pure operation (economic dispatch, unit commitment) to capacity planning (transmission and generation expansion) and to a network decision (optimal transmission switching).

Note: these "problems" are what the model solves. They are different from the "sweep modes" in notebooks 5.x, which are about running many cases at once.
