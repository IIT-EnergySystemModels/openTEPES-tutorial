# Sweep Mode B — In-memory overlay sweep

**Mode B (in-memory)** reads one baseline case into memory once, then runs each sweep case from a cheap *overlay* on top of it. Use it when your cases share most of their inputs and differ only by a parameter.

This is the natural next step after [Mode A](5.1-Sweep-Mode-A-PreBuild.ipynb), which re-reads every case folder from disk.

## The question

Same question as before, now as a real sweep from **one** folder: how does total system cost change as electricity demand grows by 0%, 10%, and 20%?

```python
import os, shutil
import pandas as pd

def coarse_copy(src, dst):
    """Copy a case folder and set a coarse time resolution so the example runs fast."""
    if os.path.exists(dst):
        shutil.rmtree(dst)
    shutil.copytree(src, dst)
    p = os.path.join(dst, "oT_Data_Parameter_9n.csv")
    df = pd.read_csv(p)
    df.loc[:, "TimeStep"] = 24   # coarse resolution, just to keep this tutorial quick
    df.to_csv(p, index=False)

# One baseline folder is enough for Mode B
coarse_copy("9n", "sweepB/9n")
print("Baseline folder is ready.")
```

## Run the sweep

An `overlay` maps a data-table name (here `"Demand"`) to a change. The simplest change is a number, which scales every value in that table. The baseline is read once; each overlay case re-builds and re-solves from it.

Each case needs its own `out_path` so the results do not collide.

```python
from openTEPES import openTEPES_Runner, openTEPES_Cases

cases = [
    openTEPES_Cases.Case("sweepB", "9n", out_path="sweepB/out_100", label="demand_100"),
    openTEPES_Cases.Case("sweepB", "9n", out_path="sweepB/out_110", label="demand_110", overlay={"Demand": 1.10}),
    openTEPES_Cases.Case("sweepB", "9n", out_path="sweepB/out_120", label="demand_120", overlay={"Demand": 1.20}),
]

records = openTEPES_Runner.run(
    cases, "appsi_highs",
    mode="in-memory", backend="serial",
    pIndOutputResults=0, pIndLogConsole=0,
)
```

```python
pd.DataFrame(records)[["label", "status", "total_cost_meur"]]
```

## What just happened

The baseline was read from disk only once. The `demand_100` case has no overlay, so it reproduces the baseline; the other two scale demand up and cost rises with it.

**Overlays can be more than a number.** An overlay value can be:
- a number — scales every value in the table,
- a function `df -> df` — for a custom change (for example, raise demand only at peak hours),
- a replacement DataFrame — swap the table entirely.

The same `backend="multiprocessing"` / `"joblib"` and `aggregate_to=...` options from Mode A apply here too.
