# Sweep Modes

A *sweep* runs many cases in one go and collects the results, instead of running them one by one. openTEPES has three sweep modes, from the most general to the fastest. The notebooks below introduce them step by step, all on the same question (how does cost change as demand rises) so you can compare the three approaches.

These sweep modes are about *running* cases. They are different from the planning and operation problems in notebook 04, which are about *what* the model solves.

## 1. Mode A — Pre-build sweep

Reads each case from its own input folder and builds and solves it independently. Use it when your cases are different folders or scenarios.

- [5.1-Sweep-Mode-A-PreBuild.ipynb](5.1-Sweep-Mode-A-PreBuild.ipynb)

## 2. Mode B — In-memory overlay sweep

Reads one baseline case into memory once, then runs each case from a cheap overlay on top of it. Use it when your cases share most of their inputs and differ by a parameter.

- [5.2-Sweep-Mode-B-InMemory.ipynb](5.2-Sweep-Mode-B-InMemory.ipynb)

## 3. Mode C — Post-build hot-swap re-solve

Builds the model once and re-solves it many times, swapping mutable parameter values in place. It is the fastest because it skips the rebuild.

- [5.3-Sweep-Mode-C-Resolve.ipynb](5.3-Sweep-Mode-C-Resolve.ipynb)
